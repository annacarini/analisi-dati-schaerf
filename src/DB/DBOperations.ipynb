```python
import pandas as pd
```

```python
# Elenco anni
listaAnni = [i for i in range(2000,2025)]

# Nomi file
filesPaths = [f"./FilesUnitiCSV/completo_{anno}.csv" for anno in listaAnni]

files = [pd.read_csv(
      filename,
      dtype={'Fascia': str, 'Cognome e Nome': str, 'Genere': str, 'Ateneo': str, 'Facoltà': str, 'S.S.D.': str, 'S.C.': str, 'Struttura di afferenza': str, 'year': str, 'area': str}
    ) for filename in filesPaths]


print("loaded")

values_fascia = []
values_ateneo = []
values_facolta = []
values_ssd = []
values_sc = []
values_struttura_afferenza = []

i = 0
for file in files:
  print(f"file {i}")
  for index, row in file.iterrows():
    # fascia
    fascia = str(row['Fascia'])
    if (fascia != "") and (not fascia.lower() in [f.lower() for f in values_fascia]):
      values_fascia.append(row['Fascia'])
    # ateneo
    ateneo = str(row['Ateneo'])
    if (ateneo != "") and (not ateneo.lower() in [a.lower() for a in values_ateneo]):
      values_ateneo.append(ateneo)
    # facolta
    facolta = str(row['Facoltà'])
    if (facolta != "") and (not facolta.lower() in [f.lower() for f in values_facolta]):
      values_facolta.append(facolta)
    # ssd
    ssd = str(row['S.S.D.'])
    if (ssd != "") and (not ssd.lower() in [s.lower() for s in values_ssd]):
      values_ssd.append(ssd)
    # sc
    sc = str(row['S.C.'])
    if (sc != "") and (not sc.lower() in [s.lower() for s in values_sc]):
      values_sc.append(sc)
    # struttura
    struttura = str(row['Struttura di afferenza'])
    if (struttura != "") and (not struttura.lower() in [s.lower() for s in values_struttura_afferenza]):
      values_struttura_afferenza.append(struttura)
  i += 1
```

```python
print("fascia:")
for v in sorted(values_fascia):
  print(v)

print("ssd:")
for v in sorted(values_ssd):
  print(v)

print("sc:")
for v in sorted(values_sc):
  print(v)
```

Per inserire il valore di area prendendolo dallo SSD

```python
anno = 2023
file_df = pd.read_csv(
      f"./FilesUnitiCSV/completo_{anno}.csv",
      dtype={'Fascia': str, 'Cognome e Nome': str, 'Genere': str, 'Ateneo': str, 'Facoltà': str, 'S.S.D.': str, 'S.C.': str, 'Struttura di afferenza': str, 'year': str, 'area': str}
    )

rows_without_ssd = []
rows_modified_counter = 0

for index, row in file_df.iterrows():
    if pd.isna(row['area']):
      if (pd.isna(row['S.S.D.'])):
         rows_without_ssd.append(index)
      else:
        row['area'] = getAreaBySSD(row['S.S.D.'])
        rows_modified_counter += 1

file_df.to_csv(f"./FilesUnitiCSV/completo_{anno}.csv",
               columns=['Fascia','Cognome e Nome','Genere','Ateneo','Facoltà','S.S.D.','S.C.','Struttura di afferenza','year','area'],
               index=False)

print(str(anno))
print(f"modificate {rows_modified_counter} righe")
print("righe senza ssd:")
print(rows_without_ssd)
```

```python
anno = 2024
file_df = pd.read_csv(
      f"./FilesUnitiCSV/completo_{anno}.csv",
      dtype={'Fascia': str, 'Cognome e Nome': str, 'Genere': str, 'Ateneo': str, 'Facoltà': str, 'S.S.D.': str, 'S.C.': str, 'Struttura di afferenza': str, 'year': str, 'area': str}
    )

rows_without_sc = []
rows_modified_counter = 0
rows_modified = []

for index, row in file_df.iterrows():
    # se non c'e' l'area
    if pd.isna(row['area']):

      # se non c'e' lo SC
      if (pd.isna(row['S.C.'])):
        #rows_without_sc.append(index)
        if index < 3714:
          area = "01 - Scienze matematiche e informatiche"
        elif index < 6584:
          area = "02 - Scienze fisiche"
        elif index < 9943:
          area = "03 - Scienze chimiche"
        elif index < 11142:
          area = "04 - Scienze della Terra"
        elif index < 16686:
          area = "05 - Scienze biologiche"
        elif index < 26659:
          area = "06 - Scienze mediche"
        elif index < 30105:
          area = "07 - Scienze agrarie e veterinarie"
        elif index < 34249:
          area = "08 - Ingegneria civile ed architettura"
        elif index < 41758:
          area = "09 - Ingegneria industriale e dell'informazione"
        elif index < 47089:
          area = "10 - Scienze dell'antichità, filologico-letterarie e storico-artistiche"
        elif index < 52250:
          area = "11 - Scienze storiche, filosofiche, pedagogiche e psicologiche"
        elif index < 57353:
          area = "12 - Scienze giuridiche"
        elif index < 63241:
          area = "13 - Scienze economiche e statistiche"
        else:
          area = "14 - Scienze politiche e sociali"
            
      # se c'e' lo SC
      else:
        area = getAreaBySSD(row['S.C.'])

      row['area'] = area

      rows_modified_counter += 1
      rows_modified.append(index)

file_df.to_csv(f"./FilesUnitiCSV/completo_{anno}.csv",
               columns=['Fascia','Cognome e Nome','Genere','Ateneo','Facoltà','S.S.D.','S.C.','Struttura di afferenza','year','area'],
               index=False)

print(str(anno))
print(f"modificate {rows_modified_counter} righe")
print("righe modificate:")
print(rows_modified)
```
